```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np . array ([50 , 60 , 70 , 80 , 90 , 100 , 110 , 120])
```

```python
y = np . array ([150 , 180 , 195 , 230 , 260 , 290 , 325 , 355])
```

```python
m = X. size
X_b = np . c_ [ np . ones ((m , 1) ) , X ]
```

```python
theta = np . linalg . inv ( X_b .T @ X_b ) @ X_b . T @ y
print (" Pendiente :", round ( theta [1] , 2) )
print (" Intercepto :", round ( theta [0] , 2) )
```

```python
precio = theta [0] + theta [1] * 95
print (" Precio estimado :", round ( precio , 2) , " miles ")
```

```python
y_pred = theta[0] + theta[1] * X

plt.scatter(X, y, color='red', label='Datos reales')
plt.plot(X, y_pred, label=f'y = {round(theta[1], 2)}x + {round(theta[0], 2)}', color='blue', linewidth=2)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Regresión Lineal')
plt.legend()
plt.show()
```
